# Reproduce the paper

Runs the full analysis from the de-identified data in `data/`, checks the outputs against the values reported in the paper, and displays Figures 1–6.

```python
import sys, runpy, pathlib
ROOT = pathlib.Path.cwd()
while not (ROOT / 'analysis' / 'run_all.py').exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / 'analysis'))
runpy.run_path(str(ROOT / 'analysis' / 'run_all.py'), run_name='__main__')
runpy.run_path(str(ROOT / 'tests' / 'test_reproduces_paper.py'), run_name='__main__')
```

```python
from IPython.display import Image, display
for f in sorted((ROOT / 'figures').glob('figure*.png')):
    print(f.name)
    display(Image(str(f), width=700))
```
